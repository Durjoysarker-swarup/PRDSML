# Logistic Regression

## Import the libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset = pd.read_csv("Social_Network_Ads.csv")
X = dataset.iloc[:, :-1].values
y = dataset.iloc[:, -1].values
```

## Spliting the datset into training set and test set

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.25, random_state = 0)
```

## Feature Scaling

Feature scaling is must for the logistic regression

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

## Tranining the logistic Regression model on the training set

```python
from sklearn.linear_model import LogisticRegression
classifier = LogisticRegression(random_state = 0)
classifier.fit(X_train, y_train)
```

## Predicting a new result

```python
info = [[30, 40000]]
print(classifier.predict(sc.transform(info)))
```

## Predicting the probability

```python
print(classifier.predict_proba(sc.transform(info)))
```

## Predicting the test set

```python
y_pred = classifier.predict(X_test)
```

```python
## concatinate to better visualization
y_pred = y_pred.reshape(len(y_pred), 1)
y_test = y_test.reshape(len(y_test), 1)
np.concatenate((y_pred, y_test), axis = 1)
```

## Making the confusion matrix and accuaracy score

```python
from sklearn.metrics import confusion_matrix, accuracy_score
cm = confusion_matrix(y_test, y_pred)
print(cm)
accuracy_score(y_test, y_pred)
```
