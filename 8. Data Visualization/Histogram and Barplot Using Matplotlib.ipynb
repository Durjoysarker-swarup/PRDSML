# Histogram and Barplot in Python Using Matplotlib


## Histogram


### Importing the libraries

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

### Returning sample from the "standard normal distribution"/"z distribution"

Standard normal distribution / z distribution
* mean 0
* Std 1

```python
np.random.seed(42)
data = np.random.randn(1000)
```

```python
print(np.std(data))
print(np.mean(data))
```

### Ploting Basic Histogram

```python
plt.hist(data, edgecolor = "black", bins = 30, color = 'skyblue')
plt.title("Basic Histogram")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

### Customized Histogram in Matplotlib with density plot

```python
sns.histplot(data, bins = 30, color = "green", kde = True)
plt.title("Customized Histogram")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

## Barplot

In case of catagorical item

```python
learners = ["Undergraduate", "Masters", "PhD", "Teachers", "Researchers"]
count = [145, 98, 50,80, 60]
```

```python
plt.bar(learners, count, color = "black", width = 0.9)
plt.title("Barplot")
plt.xlabel("Learners")
plt.ylabel("Count")
plt.show()
```

### Creating horizontal Bar plot

```python
plt.barh(learners, count, color = "black", height = 0.6)
plt.title("Barplot")
plt.xlabel("Count")
plt.ylabel("Learners")
plt.show()
```
