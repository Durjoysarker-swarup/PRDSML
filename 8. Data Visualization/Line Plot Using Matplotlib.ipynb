# LIne plot

## Importing required libraries


```python
import numpy as np
import matplotlib.pyplot as plt
```

## Defining data values

```python
x = np.array([1,2,3,4])
y = x*2
```

## Matplotlib simple line plot

```python
plt.plot(x,y)
plt.title("Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

## LIne chart with annotations

```python
from matplotlib.lines import lineStyles
# Sample data
a = [1,2,3,4,5,8]
b = [2,4,6,8,10,12]

# Creating a line chart
plt.figure(figsize=(4,3))
plt.plot(a,b, marker="o", linestyle = '-.')


# Add annotation
for i, (xi,yi) in enumerate(zip(a,b)):
    plt.annotate(f"({xi},{yi})", (xi,yi), textcoords="offset points", xytext=(0,10), ha="right")

plt.title("Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)
plt.show()
```

## Multiple plot on the same axis

```python
x = [1,2,3,4]
y = [2,4,6,8]

#first plot with x and y data
plt.plot(x,y, marker = "o")

# Add annotation
for i, (xi,yi) in enumerate(zip(x,y)):
    plt.annotate(f"({xi},{yi})", (xi,yi), textcoords="offset points", xytext=(0,10), ha="right")

a = [2,4,6,8]
b = [3,5,7,9]

plt.plot(a,b, linestyle="-.", marker = "o")

# Add annotation
for i, (xi,yi) in enumerate(zip(a,b)):
    plt.annotate(f"({xi},{yi})", (xi,yi), textcoords="offset points", xytext=(0,10), ha="right")

plt.title("Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

## Fill the area between two line

```python
x = [1,2,3,4]
y = [2,4,6,8]

#first plot with x and y data
plt.plot(x,y, marker = "o")

# Add annotation
for i, (xi,yi) in enumerate(zip(x,y)):
    plt.annotate(f"({xi},{yi})", (xi,yi), textcoords="offset points", xytext=(0,10), ha="right")

x = [1,2,3,4]
b = [3,5,7,12]

plt.plot(x,b, linestyle="-.", marker = "o")

# Add annotation
for i, (xi,yi) in enumerate(zip(x,b)):
    plt.annotate(f"({xi},{yi})", (xi,yi), textcoords="offset points", xytext=(0,10), ha="right")

plt.title("Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

plt.fill_between(x,y,b, color="green", alpha=0.2)

plt.show()
```
