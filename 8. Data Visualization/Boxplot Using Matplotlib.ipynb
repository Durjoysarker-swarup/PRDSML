## Importing the library

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
#np.random.normal(mean, sd, number or data)
np.random.seed(42)
data = np.random.normal(100, 20, 200)
```

```python
print(data)
```

```python
print(np.mean(data))
print(np.std(data))
```

## Data summarization using pandas

Pandas use on the dataframe. That's why we need to convert our data into dataframe.

```python
df = pd.DataFrame(data)
```

```python
df.describe()
```

* Boxplot(whichcar plot) visualize these distrubution( min, 25%, 50%, 75%, max)
* Identify the outliers -> if any value is not in the boxplot then the value will be outliers.


## Creating boxplot

```python
#plt.figure(figsize= (Weidth, height))
plt.figure(figsize=(5,3))
plt.boxplot(df)
plt.show()
```

image.png

- 1.5 boxdistance outside values are outliers.
- 3 times boxdistance outside are the extrime one and need to omit.

## Multiple boxplot

```python
np.random.seed(10)

d_1 = np.random.normal(100, 10, 200)
d_2 = np.random.normal(90, 20, 200)
d_3 = np.random.normal(80, 30, 200)
d_4 = np.random.normal(70, 40, 200)
d = [d_1, d_2, d_3, d_4]

plt.figure(figsize =(10, 7))
plt.boxplot(d)

plt.show()
```
